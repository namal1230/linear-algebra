```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
plt.rcParams.update({'font.size': 14, 'figure.figsize': (6,5)})
```

```python
def plotUnitBall(n, p):
    """ plot unit ball"""
    x = (2 * (np.random.rand(2,n) - 0.5))
    ind = []
    for i in range(n):
        if np.linalg.norm(x[:,i],p) <=1:
            ind.append(i)

    plt.fill(x[0, ind], x[1, ind], 'go')

    plt.xlabel("$x_1$")
    plt.ylabel("$x_2$")
    plt.grid()
    plt.show()
```

```python
plotUnitBall(n=50000, p=2)
```

```python
plotUnitBall(n=100000, p=1)
```
