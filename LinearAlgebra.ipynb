```python
import numpy as np
import matplotlib.pyplot as plt

# Day 1 = April 1st
n = np.arange(1, 61)  # 60 days out

# Mark I: 1 kJ/day starting day 1
E1 = 1 * n

# Mark II: 4 kJ/day starting day 31 (May 1st)
E2 = np.where(n >= 31, 4 * (n - 30), 0)

# Find intersection (analytical): n = 4(n-30) -> n = 40
intersect_day = 40
intersect_energy = intersect_day  # E1(40) = 40

plt.figure(figsize=(9, 6))
plt.plot(n, E1, label="Mark I (1 kJ/day)", linewidth=2, color="#1f77b4")
plt.plot(n, E2, label="Mark II (4 kJ/day, starts day 31)", linewidth=2, color="#ff7f0e")

plt.scatter([intersect_day], [intersect_energy], color="red", zorder=5)
plt.annotate(f"Day {intersect_day} (May 10)\n{intersect_energy} kJ each",
             xy=(intersect_day, intersect_energy),
             xytext=(intersect_day + 3, intersect_energy - 15),
             arrowprops=dict(arrowstyle="->", color="red"))

plt.axvline(31, color="gray", linestyle="--", alpha=0.5, label="May 1 (Mark II starts)")

plt.xlabel("Day (1 = April 1st)")
plt.ylabel("Total energy generated (kJ)")
plt.title("Cumulative Energy: Mark I vs Mark II Solar Panels")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
print(f"Intersection at day {intersect_day}, energy {intersect_energy} kJ each, total {2*intersect_energy} kJ")
```
